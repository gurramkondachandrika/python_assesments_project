```python
'''2.Count Characters
Problem

Take a sentence from the user and print:

Total number of characters
Total number of words
Example

Input:

Python is easy

Output:

Characters: 14
Words: 3
'''
sen = "Take a sentence from the user and print"
Total_number_of_characters = len(sen)
Total_number_of_words = len(sen.split())
print(f"Total number of characters : {Total_number_of_characters}")
print(f"Total number of words:  {Total_number_of_words}")


```

```python
'''
6. Mask Bank Account Number
Problem

Take a bank account number and hide all digits except last 4 digits.

Example

Input:

123456789012

Output:

XXXXXXXX9012

'''
account_num = input("enter your account number:")
new_out = "X" * (len(account_num)-4) + account_num[-4:]
print(new_out)
```

```python
'''
9. Email Validator
Problem

Take an email from user and:

Convert to lowercase
Remove extra spaces
Check if @gmail.com exists
Example

Input:

   AMIT123@GMAIL.COM

Output:

Valid Gmail Address
'''
email = input("Take an email from user and:")
#print(email.lower())
#print(email.strip())
email = email.lower().strip()
if "@gmail.com" in email:
  print("valid email")
else:
  print("Invalid")

```
