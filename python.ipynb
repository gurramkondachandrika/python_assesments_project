```python
'''

31.Password Strength Checker
Problem

Check whether password:

Has uppercase
Has lowercase
Has numbers
Minimum 8 characters

'''
password = input("enter your password: ")
has_upper = True
has_lower = True
has_number = True
for char in password:
  if char.isupper():
    has_upper = True
  elif char.islower():
    has_lower = True
  elif char.isdigit():
    has_number = True
print("has Uppercase:",has_upper)
print("has Uppercase:",has_lower)
print("has Uppercase:",has_number)
if has_upper and has_lower and has_number:
  print("strong password")
else:
  print("weak password")
```
